# Convolutional Neural Networks (CNN) - The Ultimate Deep Guide

අපි මේ වෙනකන් කතා කළේ සාමාන්‍ය Artificial Neural Networks (ANN) නැත්නම් Dense layers ගැන. හැබැයි අපි පරිගණකයට පින්තූර (Images) තේරුම් ගන්න හදද්දී මේ සාමාන්‍ය ANN එක සම්පූර්ණයෙන්ම අසාර්ථක වෙනවා. 

මේකට විසඳුමක් විදිහට තමයි **CNN (Convolutional Neural Networks)** කරලියට එන්නේ. අද කාලේ Computer Vision (පරිගණක දෘෂ්ටිය) වගේම Self-Driving Cars, MRI ස්කෑන් බලලා ලෙඩ හොයාගන්න එක, Facial Recognition මේ හැමදේකම පදනම තමයි CNN.

අපි මේ Notebook එකෙන් CNN කියන්නේ මොකක්ද, ඒක ඇතුලේ තියෙන ගණිතය වැඩ කරන්නේ කොහොමද, පින්තූරයක් ඇතුලට දුන්නම ඒක තේරුම් ගන්නේ කොහොමද කියන එක ගැන කිසිම කම්මැලිකමක් නැතුව, අතිශය ගැඹුරින් (deeply) කතා කරමු.

## 1. Human Brain Vs CNN (මොළය සහ CNN අතර පුදුම හිතෙන සම්බන්දය)

CNN කියන සංකල්පය නිකම්ම කවුරුහරි ගණිතඥයෙක් පරිගණකයක් ඉස්සරහා ඉඳගෙන හදපු එකක් නෙවෙයි. මේක හැදුණේ අපේ ඇහැ සහ මොළය පින්තූර තේරුම් ගන්න විදිහ (Visual Cortex) ගැන කරපු ජීව විද්‍යාත්මක පර්යේෂණ වලින්.

1959 දී Hubel සහ Wiesel කියන විද්‍යාඥයෝ දෙන්නා බළලෙකුගේ මොළයට වයර් සම්බන්ධ කරලා පර්යේෂණයක් කළා. එයාලා හිතුවේ බළලට මුහුණක් පෙන්නුවාම මුළු මොළයම එකපාර වැඩ කරලා "මේක මුහුණක්" කියලා තේරුම් ගනියි කියලා. හැබැයි පුදුමයකට වගේ එහෙම වුණේ නෑ.

එයාලා හොයාගත්තා මොළයේ තියෙන එක එක නියුරෝන ක්‍රියාත්මක වෙන්නේ හරිම නිශ්චිත දේවල් වලට විතරයි කියලා (මේකට Receptive Fields කියලා කියනවා).
- සමහර නියුරෝන ක්‍රියාත්මක වුණේ තිරස් ඉරි (horizontal lines) දැක්කාම විතරයි.
- තව සමහර ඒවා ක්‍රියාත්මක වුණේ සිරස් ඉරි (vertical lines) දැක්කාම විතරයි.

**එතකොට අපේ මොළය පින්තූරයක් තේරුම් ගන්නේ කොහොමද?**
මේක වෙන්නේ පියවරෙන් පියවර (Hierarchical විදිහට):
1. **පළවෙනි Layers:** ඇහැ දකිනවා පින්තූරයේ තියෙන පොඩි පොඩි ඉරි සහ මායිම් (Edges / Lines).
2. **මැද Layers:** ඒ ඉරි එකතු කරලා හැඩතල (Shapes, Corners, රවුම්) හදාගන්නවා.
3. **අන්තිම Layers:** ඒ හැඩතල ඔක්කොම එකතු කරලා "ආ මේක බල්ලෙක්ගේ මුහුණක්" නැත්නම් "මේක කාරෙකක්" කියලා සම්පූර්ණ රූපය තේරුම් ගන්නවා.

CNN එකෙත් වෙන්නේ හරියටම මේ දේම තමයි (Hierarchical Feature Extraction). අපි දාන මුල්ම Convolution layers වලින් පින්තූරයේ තියෙන පොඩි ඉරි හොයනවා. ඊළඟට දාන layers වලින් ඒ ඉරි පාවිච්චි කරලා හැඩතල හොයනවා. අන්තිමට තියෙන Dense Layers වලින් සම්පූර්ණ රූපය මොකක්ද කියලා තීරණය කරනවා.

## 2. All you need to Know about Images (පරිගණකයට පින්තූරයක් පේන්නේ කොහොමද?)

අපිට පින්තූරයක් ලස්සනට පෙනුණට, පරිගණකයකට පින්තූරයක් කියන්නේ නිකම්ම ඉලක්කම් ගොඩක් පිරිච්ච කොටු දැලක් (Matrix of numbers) විතරයි.

### Grayscale Images (කළු සුදු පින්තූර)
මේකෙදී පින්තූරය හැදිලා තියෙන්නේ එකම එක 2D Matrix එකකින් (දිග x පළල). මේකේ තියෙන හැම කොටුවකටම අපි කියනවා Pixel (පික්සල්) එකක් කියලා. මේ Pixel එකක අගය (8-bit පින්තූරයක) 0 ඉඳන් 255 වෙනකන් තියෙන්න පුළුවන්. 
- 0 කියන්නේ ආලෝකය නෑ (සම්පූර්ණයෙන්ම කළු).
- 255 කියන්නේ උපරිම ආලෝකය (සම්පූර්ණයෙන්ම සුදු).
මේ අතර තියෙන අගයන් අළු පාටේ විවිධ ඡායාවන් වෙනවා (උදා: 128 කියන්නේ අළු පාට).

### Color Images (RGB - වර්ණ පින්තූර)
වර්ණ පින්තූරයකදී එක Matrix එකක් නෙවෙයි, Matrices 3ක් (Layers 3ක්) එක උඩ එක තියලා (Stack කරලා) තමයි පින්තූරය හැදෙන්නේ. ඒවට අපි කියනවා **Channels** කියලා. ලෝකයේ තියෙන ඕනෑම පාටක් මේ ප්‍රාථමික වර්ණ 3න් හදන්න පුළුවන්:
1. Red Channel (රතු)
2. Green Channel (කොළ)
3. Blue Channel (නිල්)

උදාහරණයක් විදිහට පින්තූරයේ ප්‍රමාණය 28x28 (උස x පළල) නම්:
- කළු සුදු පින්තූරයක හැඩය (Shape): `(28, 28, 1)`
- වර්ණ පින්තූරයක හැඩය: `(28, 28, 3)`

![RGB Channels Separation](https://cs231n.github.io/assets/cnn/depthcol.jpeg)

මේ පින්තූරයෙන් පැහැදිලිව පේනවා වර්ණ පින්තූරයක් කියන්නේ අර Red, Green, Blue කියන 2D Matrices තුන එකතු වෙලා හැදුණ 3D බ්ලොක් (Volume) එකක් කියලා. අපි CNN එකකට input එක දෙද්දී දෙන්නෙත් මේ 3D volume එකමයි.

## 3. Operation Of CNN Vs ANN (ANN එක පින්තූර වලට ෆේල් වෙන්නේ ඇයි?)

ඇයි අපිට මේ තරම් අලුත් CNN එකක් ඕනේ? කලින් පාවිච්චි කරපු Artificial Neural Network (ANN) එකම පින්තූර වලට පාවිච්චි කරන්න බැරිද?

උත්තරේ තමයි බෑ. ලොකුම ප්‍රශ්න 2ක් තියෙනවා:

### ප්‍රශ්නය 1: සම්බන්ධතා නැතිවීම (Loss of Spatial Information)
ANN එකකට (Dense layers වලට) දත්ත දෙන්න නම්, අපි 2D පින්තූරය දිග තනි පේළියකට කඩන්න (Flatten කරන්න) ඕනේ. 
හිතන්න ඔයාගේ මුහුණේ පින්තූරයක් තියෙනවා. ඇහැක් තියෙන්නේ නහයට උඩින්. හැබැයි මේ පින්තූරය තනි පේළියකට කැඩුවාම, ඒ "උඩින් තියෙනවා", "පැත්තෙන් තියෙනවා", "වටේට තියෙනවා" කියන සම්බන්ධය සම්පූර්ණයෙන්ම නැති වෙලා යනවා. පින්තූරයක අර්ථය හැදෙන්නේ පික්සල් එකට එකක් ළඟින් තියෙන විදිහ අනුවයි මිසක් තනි පේළියක නෙවෙයි.

### ප්‍රශ්නය 2: Parameters ප්‍රමාණය අතිවිශාල වීම (The Parameter Explosion)
හිතන්න ඔයාගේ පින්තූරය අද කාලේ තියෙන සාමාන්‍ය 1080x1080 (HD) ප්‍රමාණයේ වර්ණ (RGB) පින්තූරයක් කියලා. එතකොට Input Layer එකේ විතරක් Neurons (1080 x 1080 x 3) = **මිලියන 3.5ක් විතර** තියෙනවා!
ඔයා ඊළඟ Hidden Layer එකට Neurons 1000ක් විතරක් දැම්මොත්, පළවෙනි Hidden layer එකට විතරක් Weights **බිලියන 3.5ක්** ඕනේ වෙනවා! මේ තරම් Parameters ප්‍රමාණයක් Train කරනවා කියන්නේ Memory එකට දරාගන්න බැරි, සති ගාණක් යන වැඩක්. Computer එක Crash වෙන එක විතරයි වෙන්නේ.

**CNN වල විසඳුම:** 
CNN මගින් පින්තූරය තනි පේළියකට කඩන්නේ නෑ. ඒකෙ 2D/3D හැඩය ඒ විදිහටම තියාගෙන **"Filter"** එකක් පාවිච්චි කරලා කොටසින් කොටස (patch by patch) බලනවා. ඒක නිසා Spatial Information නැති වෙන්නෙත් නෑ, Weights ගාණ බිලියන ගණන් වලින් දහස් ගණනකට අඩු වෙනවා. මේ සංකල්පයට කියන්නේ **Weight Sharing** (එකම filter එක පින්තූරය පුරාම පාවිච්චි කරන නිසා) සහ **Local Connectivity** (එකපාරට බලන්නේ පින්තූරෙන් පොඩි කොටසක් විතරයි) කියලා.

## 4. Convolution Operation In CNN (ගණිතමය විග්‍රහය)

Convolution කියන්නේ CNN වල තියෙන හදවත. මේකෙන් කරන්නේ පින්තූරයේ තියෙන වැදගත් ලක්ෂණ (Features - වගේ ඉරි, හැඩතල) උරා ගන්න එක (Feature Extraction).

**මේක වැඩ කරන්නේ කොහොමද?**
හිතන්න ඔයා ලොකු පින්තූරයක් දිහා පොඩි විශාලන කාචයකින් (Magnifying glass) කොටසින් කොටස බලනවා කියලා. 
මේ කාචයට අපි කියනවා **Kernel** එකක් නැත්නම් **Filter** එකක් කියලා. සාමාන්‍යයෙන් මේවා 3x3 ප්‍රමාණයේ පොඩි කොටු. (Deep Learning වලදී Network එක ඉගෙනගන්න Weights කියන්නේ අන්න මේ 3x3 Filter එකේ ඇතුලේ තියෙන ඉලක්කම් 9 යි!).

1. **Scan කිරීම:** මේ 3x3 Filter එක පින්තූරයේ වම් පැත්තේ උඩම කෙළවරින් තියනවා.
2. **Dot Product (ගුණ කරලා එකතු කිරීම):** පින්තූරයේ තියෙන පික්සල් අගයන් 9 සහ Filter එකේ තියෙන අනුරූප ඉලක්කම් 9 එකිනෙකාට ගුණ කරනවා. ඊටපස්සේ ඒ ගුණිතයන් ඔක්කොම එකතු කරනවා. (උදා: `(1*1) + (0*0) + (1*1) ...` = X). අන්තිමට එන්නේ තනි ඉලක්කමක්.
3. **Feature Map:** ඒ එන උත්තරේ අලුත් Matrix එකක පළවෙනි කොටුවේ ලියනවා.
4. **Stride (පියවර):** ඊටපස්සේ Filter එක එක පියවරක් දකුණට අරන් ගිහින් කලින් දේම කරනවා. මේ අරන් යන පියවරක දුරට අපි කියනවා **Stride** කියලා. Stride එක 1යි නම් එක කොටුවක් යනවා. 2යි නම් කොටු 2ක් පැනලා යනවා (එතකොට output පින්තූරය ගොඩක් පොඩි වෙනවා).
5. මේ විදිහට මුළු පින්තූරයම කොටසින් කොටස scan කරනවා.

![2D Convolution Animation](https://upload.wikimedia.org/wikipedia/commons/1/19/2D_Convolution_Animation.gif)

මේ විදිහට Scan කළාම හැදෙන අලුත් පින්තූරයට අපි කියනවා **Feature Map** එකක් (නැත්නම් Activation Map එකක්) කියලා. 
එක Filter එකක් එක ලක්ෂණයක් (උදා: තිරස් ඉරි) විතරයි හොයන්නේ. අපි Model එකට කියනවා "Filters 32ක් පාවිච්චි කරන්න" කියලා. එතකොට Model එකම අර ඉලක්කම් 9 (Weights) එක එක විදිහට adjust කරගෙන (Backpropagation මගින්), එක එක Filters වලින් එක එක ජාතියේ ලක්ෂණ (ඉරි, රවුම්, හැඩතල) හොයාගන්න ඉගෙන ගන්නවා.

## 5. Padding (භාවිතා කරන ආකාරය සහ ගැඹුරු පැහැදිලි කිරීම)

අර උඩ තියෙන Convolution Operation එක කරද්දී ලොකු ප්‍රශ්න දෙකක් එනවා. 

1. **පින්තූරය කුඩා වීම (Shrinking):** 5x5 පින්තූරයකට 3x3 filter එකක් දැම්මම Output එක 3x3 වෙනවා. අපි Layers 10ක් විතර දැම්මොත් අන්තිමට පින්තූරය 1x1 වෙලා නැතිවෙලාම යයි!
2. **කොන් වල දත්ත නැතිවීම (Information loss at borders):** පින්තූරයේ කොන් වල (edges) තියෙන පික්සල් පාවිච්චි වෙන්නේ (Filter එකට අහුවෙන්නේ) එක පාරක් විතරයි. හැබැයි මැද තියෙන පික්සල් කීප සැරයක්ම Filter එකට අහුවෙනවා. ඒ කියන්නේ පින්තූරයේ මායිම් වල තියෙන වැදගත් දත්ත Network එකට මගහැරෙනවා.

**විසඳුම: Padding**
Padding කියන්නේ පින්තූරයේ වටේට බිංදු (Zeros) වලින් රාමුවක් (Border එකක්) අඳින එකටයි. හරියට පින්තූරයකට frame එකක් දැම්මා වගේ.

ගණිතමය සූත්‍රය: `Output Size = [(W - K + 2P) / S] + 1`
(W = Input size, K = Kernel size, P = Padding, S = Stride).
අපි Padding (P) එකක් දුන්නොත් Output size එක අඩුවෙන්නේ නැතුව සමතුලිත වෙනවා.

**Keras වල Padding වර්ග දෙකයි:**
1. **Valid Padding (`padding='valid'`):** මුකුත් කරන්නේ නෑ (Zeros දාන්නේ නෑ). මේක තමයි default එක. Output එකේ ප්‍රමාණය අඩු වෙනවා.
2. **Same Padding (`padding='same'`):** Output එකේ ප්‍රමාණයයි Input එකේ ප්‍රමාණයයි සමාන (Same) වෙන විදිහට පරිගණකය විසින්ම වටේට Zeros දානවා. මේකෙන් පින්තූරය කුඩා වීමත් නවතිනවා, කොන් වල දත්ත නැතිවෙලා යන එකත් නවතිනවා.

```python
# Padding භාවිතා කරන ආකාරය (Code Example)
from tensorflow.keras.layers import Conv2D
from tensorflow.keras.models import Sequential

model_pad = Sequential()

# Valid Padding (ප්‍රමාණය කුඩා වේ)
# Input 32x32 -> Output 30x30 වෙනවා
model_pad.add(Conv2D(32, kernel_size=(3, 3), padding='valid', input_shape=(32, 32, 3)))

# Same Padding (ප්‍රමාණය ඒ විදිහටම තියෙනවා)
# Input 30x30 -> Output 30x30 ම වෙනවා
model_pad.add(Conv2D(32, kernel_size=(3, 3), padding='same'))

print("Padding controls how the image shrinks during Convolution!")
```

## 6. Pooling Layers (භාවිතා කරන ආකාරය සහ ඇයි Max Pooling හොඳම?)

Convolution එකෙන් අපි Features හෙව්වා. දැන් අපිට ඕනේ මේ පින්තූරයේ තියෙන **වැදගත්ම** ලක්ෂණ ටික විතරක් තියාගෙන, අනිත් අනවශ්‍ය දේවල් අතෑරලා පින්තූරයේ size එක භාගයකින් විතර පොඩි කරන්න (Downsampling). ඒකට අපි පාවිච්චි කරනවා **Pooling Layers**.

**ඇයි මේක කරන්නේ? අරමුණු 2යි:**
1. **Computation අඩු කිරීම:** ප්‍රමාණය අඩු කළාම (උදා: 32x32 ඉඳන් 16x16 ට) ඉගෙනගන්න තියෙන weights ගාණ අඩු වෙනවා. ඒක නිසා Model එක Train වෙන්න යන වෙලාව ගොඩක් ඉතුරු වෙනවා.
2. **Translation Invariance (අතිශය වැදගත්):** හිතන්න ඔයා බළලෙක්ගේ පින්තූරයක් දුන්නා. ඒ බළලා පින්තූරයේ වම් පැත්තෙ හිටියත්, දකුණු පැත්තෙ හිටියත් Model එකට ඒ බළලාමයි කියලා අඳුරගන්න පුළුවන් වෙන්න ඕනේ. Pooling වලින් කරන්නේ පොඩි ප්‍රදේශයක තියෙන ප්‍රබලම ලක්ෂණය විතරක් උරාගන්න එක නිසා, පින්තූරයේ ලක්ෂණ පොඩ්ඩක් එහා මෙහා වුණත් (Shift වුණත්) Model එකට ප්‍රශ්නයක් වෙන්නේ නෑ.

**කොහොමද කරන්නේ?**
මේකත් Convolution වගේමයි. පොඩි කවුළුවක් (උදා: 2x2) පින්තූරය උඩින් අරන් යනවා. හැබැයි මුකුත් ගුණ කරන්නේ (Weights නෑ) නෑ.

1. **Average Pooling:** 2x2 කොටුව ඇතුලේ තියෙන අගයන් 4න් සාමාන්‍යය (Average) ගන්නවා. මේකෙන් වෙන්නේ පින්තූරය බ්ලර් (Blur) වෙන එකයි. ඒක නිසා වැදගත් Features (වගේ තියුණු ඉරි) මැකිලා යනවා. මේක අද කාලේ පාවිච්චි කරන්නේම නැති තරම්.
2. **Max Pooling:** ඒ 2x2 කොටුව ඇතුලේ තියෙන ලොකුම (Maximum) අගය විතරක් ගන්නවා. මේක තමයි **99% ක්ම වෙලාවට පාවිච්චි කරන්නේ**. ඇයි? Convolution වලින් එන අගයක් ලොකුයි කියන්නේ, එතන අනිවාර්යයෙන්ම අපි හොයන Feature එක (උදා: ඇසක්, කනක්) තියෙනවා කියන එකයි. Max Pooling වලින් කරන්නේ ඒ "ප්‍රබලම සාක්ෂිය" විතරක් ඉස්සරහට අරන් යන එකයි. (හරියට කාමරයක කෑගහන මිනිස්සු 4කින්, සද්දෙන්ම කෑගහන කෙනාගේ සද්දේ විතරක් අහනවා වගේ වැඩක්).

```python
# Pooling භාවිතා කරන ආකාරය (Code Example)
from tensorflow.keras.layers import MaxPooling2D, AveragePooling2D

model_pool = Sequential()

# Convolution Layer එකකින් පස්සේ Max Pooling දාන විදිහ
model_pool.add(Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(32, 32, 3)))

# 2x2 කවුළුවක් (pool_size) අරගෙන, පියවර 2ක් (strides) පනිමින් ලොකුම අගය ගන්නවා
# මේකෙන් පින්තූරයේ උස සහ පළල හරියටම භාගයක් වෙනවා (උදා: 30x30 -> 15x15)
model_pool.add(MaxPooling2D(pool_size=(2, 2), strides=(2, 2)))

print("Max Pooling forces the network to focus only on the most important features!")
```

## 7. Flattening and Fully Connected Layers

අපි දැන් Convolution (Features හොයන්න) සහ Pooling (ප්‍රමාණය අඩු කරන්න) Layers කීපයක්ම එක පස්සේ එක (Layer by Layer) පාවිච්චි කළා කියමු. 
දැන් අන්තිමට ඉතුරු වෙලා තියෙන්නේ පින්තූරයේ තියෙන ලක්ෂණ (Features) ඔක්කොම උරාගත්ත, හැබැයි ප්‍රමාණයෙන් පොඩි 3D Matrix එකක්. 

දැන් අපිට ඕනේ මේ උරාගත්ත Features ටික පාවිච්චි කරලා මේ පින්තූරේ මොකක්ද කියලා අවසාන තීරණය ගන්න එකයි (Classification - මේක බල්ලෙක්ද, පූසෙක්ද).

මේ තීරණය ගන්න අපි යන්න ඕනේ සාමාන්‍ය Artificial Neural Network (ANN) එකක් හරහා (Dense Layers). හැබැයි අපි දන්නවා Dense layer එකකට 3D matrices දෙන්න බෑ, ඒකට ඕනේ 1D vector (තනි ඉලක්කම් පේළියක්) එකක් කියලා.

**Flattening:**
ඒ නිසා අපි අර අන්තිමට ඉතුරු වෙලා තියෙන 3D block එක (උදා: 8x8x64), දිග තනි පේළියකට කඩනවා. ඒකට තමයි Flattening කියන්නේ. (8 * 8 * 64 = අගයන් 4096 ක පේළියක්).

**Fully Connected Layers (Dense Layers):**
අර 4096 ක දිග පේළිය අපි සම්ප්‍රදායික ANN එකකට දෙනවා. මේ Neurons එකිනෙකට ඔක්කොම සම්බන්ධ වෙලා තියෙන නිසා තමයි "Fully Connected" කියන්නේ. ඒකෙ අන්තිම Layer එකට (classes 10ක් තියෙනවා නම් Neurons 10ක් දාලා) Softmax පාවිච්චි කරලා අපේ අවසාන උත්තරේ සම්භාවිතාවක් (Probability) විදිහට ගන්නවා.

## 8. Real World Project Plan (ඇත්ත Project එකකට බහිද්දී ගහන්න ඕනේ Plan එක)

කෙලින්ම කෝඩ් කරන්න කලින්, ඇත්ත ලෝකයේ CNN Project එකක් කරද්දී අපි අනිවාර්යයෙන්ම පිළිවෙලකට යන්න ඕනේ. මේ තියෙන්නේ ඒ අතිශය වැදගත් Plan එක:

### පියවර 1: Data එකතු කිරීම සහ සැකසීම (Preprocessing)
- **Image Resizing:** පින්තූර එක එක ප්‍රමාණ වලින් තියෙන්න බෑ. ඔක්කොම එකම Size එකකට (උදා: `224x224`) කපාගන්න ඕනේ.
- **Normalization:** පික්සල් වල අගයන් 0-255 අතර තියෙන්නේ. ඒ ඔක්කොම `255.0` න් බෙදලා 0 ත් 1 ත් අතරට ගේන්න ඕනේ. (එතකොට Model එකට Weights අප්ඩේට් කරන්න ලේසියි, Gradient එක stable වෙනවා).
- **Data Augmentation:** අපිට තියෙන්නේ පින්තූර 1000යි නම්, Deep Learning වලට ඒක මදි. ඒ නිසා තියෙන පින්තූරම ටිකක් හරවලා (Rotate), Zoom කරලා, අනිත් පැත්ත හරවලා (Flip) ව්‍යාජ විදිහට Dataset එක ලොකු කරගන්න ඕනේ.

### පියවර 2: Architecture එක තීරණය කිරීම (Architecture Selection)
- **මුලින්ම සරලව යන්න:** එකපාරටම සංකීර්ණ දේවල් ලියන්නේ නැතුව උඩ තියෙනවා වගේ සරල Convolution/Pooling Layers 2, 3 ක් දාලා බලන්න (Baseline model එකක් හදාගන්න).
- **Transfer Learning (අතිශය වැදගත්):** අද කාලේ කවුරුත් Scratch (මුල ඉඳන්) ඉඳන් ලොකු Models හදන්නේ නෑ. Google, Microsoft වගේ සමාගම් Super Computers පාවිච්චි කරලා මිලියන ගාණක් පින්තූර වලින් මාස ගණන් Train කරපු `VGG16`, `ResNet`, `MobileNet` වගේ සුපිරි Models තියෙනවා. අපි කරන්නේ ඒවගේ "මොළය" (ඉගෙනගත්ත Weights ටික) අපේ ප්‍රොජෙක්ට් එකට අරන් පාවිච්චි කරන එකයි. මේකට තමයි Transfer Learning කියන්නේ.

### පියවර 3: Overfitting පාලනය කිරීම
- CNN එකකට පින්තූර කටපාඩම් කරන්න හරි ලේසියි. ඒ නිසා අනිවාර්යයෙන්ම Dense layers අස්සට **Dropout Layers** (උදා: `Dropout(0.5)`) පාවිච්චි කරන්න.
- **Early Stopping:** Training වෙද්දී Validation Loss එක දිගටම අඩු වෙන්න ඕනේ. ඒක වැඩි වෙන්න ගත්තොත් (කටපාඩම් කරන්න ගත්තොත්) එතනින්ම Training එක නවත්තන්න Auto ක්‍රමයක් (Callback එකක්) සකස් කරන්න.

### පියවර 4: Evaluation (ඇගයීම)
- නිකම්ම Accuracy එක දිහා බලලා රැවටෙන්න එපා! 99% accuracy තිබ්බත් සමහරවිට model එක අසාර්ථක වෙන්න පුළුවන් (Data imbalanced නම්).
- හැමවෙලේම **Confusion Matrix**, **Precision**, සහ **Recall** වගේ metrics පාවිච්චි කරලා කොයි පන්තියද (class) හරියටම වරදින්නේ කියලා බලන්න ඕනේ.

## 9. CNN Example with RGB (මේ ඔක්කොම එකතු කරලා කෝඩ් කරමු)

අපි දැන් ඉගෙනගත්ත Convolution, Padding, Pooling, Flattening ඔක්කොම එකතු කරලා සම්පූර්ණ CNN එකක් Keras වලින් ලියමු.

මේක වැඩ කරන විදිහ:
1. **Input Image:** 32x32 ප්‍රමාණයේ RGB පින්තූරයක් (`shape = 32x32x3`).
2. **Convolution Layer 1:** 3x3 Filters 32ක් පාවිච්චි කරනවා. `padding='same'` නිසා Output එක `32x32x32` වෙනවා. ReLU වලින් සෘණ අගයන් අයින් කරනවා.
3. **Pooling Layer 1:** 2x2 Max Pooling නිසා උසයි පළලයි භාගයක් වෙනවා: `16x16x32`.
4. **Convolution Layer 2:** Filters 64ක් දානවා. Output: `16x16x64`.
5. **Pooling Layer 2:** ආයෙත් භාගයක් කරනවා. Output: `8x8x64`.
6. **Flattening:** 8x8x64 3D block එක දිග පේළියකට කඩනවා (8 * 8 * 64 = `4096`).
7. **Fully Connected Layer:** අර 4096 Neurons 128ක Dense layer එකකට දෙනවා.
8. **Output Layer:** Classes 10කට අදාළව Softmax හරහා අවසාන උත්තරේ ගන්නවා.

```python
# Mac පරිගණක වල (M1/M2/M3) TensorFlow Kernel එක Crash වෙන ප්‍රශ්නයට විසඳුමක් විදිහට CPU එක පාවිච්චි කරන්න කියලා මෙතනින් කියනවා.
import os
os.environ['CUDA_VISIBLE_DEVICES'] = '-1'

import tensorflow as tf
# GPU/Metal plugin එක අක්‍රිය කිරීම (Crash වීම නැවැත්වීමට)
try:
    tf.config.set_visible_devices([], 'GPU')
except:
    pass

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model = Sequential([
    # 1. Convolution (Filters 32, Size 3x3, Padding='same', Input shape: 32x32 size with 3 RGB channels)
    Conv2D(32, kernel_size=(3, 3), padding='same', activation='relu', input_shape=(32, 32, 3)),
    
    # 2. Max Pooling (2x2)
    MaxPooling2D(pool_size=(2, 2)),
    
    # 3. Convolution (Filters 64, Size 3x3, Padding='same')
    Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'),
    
    # 4. Max Pooling
    MaxPooling2D(pool_size=(2, 2)),
    
    # 5. Flattening (Matrix එක තනි පේළියක් කිරීම)
    Flatten(),
    
    # 6. Fully Connected Layers (අවසාන තීරණය ගන්න ANN එක)
    Dense(128, activation='relu'),
    Dense(10, activation='softmax') # Classes 10ක් තියෙනවා නම්
])

# Model එකේ සැකිල්ල බලාගන්න පුළුවන්
model.summary()
```

## 10. Understanding the Model Summary (අවසාන Output එක තේරුම් ගනිමු)

උඩ Code එක Run කළාම `model.summary()` එකෙන් ලස්සන වගුවක් එනවා. ඒකේ තියෙන දේවල් මොනවද කියලා ගැඹුරින් බලමු:

### 1. Output Shape (හැඩය)
- හැම layer එකකම output එක `(None, 32, 32, 32)` වගේ පෙන්වනවා.
- මෙතන **`None`** කියන්නේ **Batch Size** එක. අපි පින්තූර 32ක් එකපාර දෙනවද, 64ක් දෙනවද කියලා Network එක තාම දන්නේ නෑ. ඒක තීරණය වෙන්නේ `model.fit()` කරද්දී (Training වෙලාවටයි).
- අනිත් ඉලක්කම් 3 න් `(උස, පළල, Channels/Filters ගණන)` පෙන්වනවා.
- Max Pooling වලින් පස්සේ උසයි පළලයි `32x32` ඉඳන් `16x16` ට භාගයක් වෙන හැටි ඔයාට Output Shape එක දිහා බැලුවම පැහැදිලිව පෙනෙයි.

### 2. Param # (Parameters ගණන - ගණිතය)
මේ කියන්නේ Network එකට ඉගෙනගන්න තියෙන ඉලක්කම් (Weights සහ Biases) ගාණයි.
- **Conv2D:** Filters ගාණ වැඩිවෙන්න වැඩිවෙන්න ඉගෙනගන්න ඕනේ parameters ගාණ වැඩිවෙනවා. 
  *(ගණනය කිරීම: 3x3 Filter එකකට Weights 9යි, ඒකට අදාළ Bias එක 1යි. ඔක්කොම 10යි. මුලින්ම Channels 3ක් තියෙන නිසා `10 x 3 = 30`. ඒ වගේ Filters 32ක් තියෙන නිසා `30 x 32 = 960` වගේ ගාණක් එනවා).* 
- **MaxPooling & Flatten:** මේවගේ Param ගාණ **0** යි. ඇයි? මොකද මේවා නිකම්ම "ලොකුම එක තෝරනවා", නැත්නම් "දිගට කඩනවා" වගේ මෙහෙයුම් (Operations) විතරයි. එතන ඉගෙනගන්න කිසිම Weight එකක් නෑ.
- **Dense (Fully Connected):** ඔයාට පෙනෙයි අතිවිශාලම Parameters ගාණක් (ලක්ෂ ගාණක්) තියෙන්නේ මෙතන කියලා. මොකද Flatten කරපු (4096) හැම Neuron එකක්ම, ඊළඟ Layer එකේ තියෙන Neurons 128ටම වයර් වලින් සම්බන්ධ වෙනවා. 
  *(ගණනය කිරීම: `4096 inputs x 128 neurons = 524,288` Weights + `128` Biases. සම්පූර්ණ ගාණ ලක්ෂ 5කට වැඩියි!)*
