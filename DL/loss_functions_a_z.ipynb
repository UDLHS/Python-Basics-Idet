## 1. Loss Function Vs Cost Function (වෙනස කුමක්ද?)

ගොඩක් වෙලාවට මේ වචන දෙක මාරුවෙන් මාරුවට පාවිච්චි වෙනවා. හැබැයි මේ දෙක අතර පොඩි, නමුත් වැදගත් වෙනසක් තියෙනවා.

**Loss Function:**
මේකෙන් මනින්නේ **එක training example එකක් (තනි දත්තයක්)** මත model එක කරපු වැරැද්ද. 
උදාහරණයක් විදිහට, එක ගෙදරක මිල අනුමාන කරද්දී වෙන වැරැද්ද තමයි Loss එක.

**Cost Function:**
මේකෙන් මනින්නේ **සම්පූර්ණ Training Dataset එකටම (හෝ Batch එකකට)** model එක කරපු සාමාන්‍ය වැරැද්ද (Average Error එක). 
ඒ කියන්නේ ගෙවල් 1000ක මිල අනුමාන කරලා, ඒ ඔක්කොගෙම Loss වල සාමාන්‍ය අගය (Average) ගත්තාම ඒකට අපි කියනවා Cost Function එක කියලා.

සරල සමීකරණය:
Cost = (Loss_1 + Loss_2 + ... + Loss_N) / N

අපි Neural Network එකක් train කරද්දී ඇත්තටම minimize කරන්න උත්සහ කරන්නේ Cost Function එකයි. මොකද අපිට එක දත්තයකින් විතරක් model එක හරිද කියන්න බෑ, මුළු dataset එකේම average වැරැද්ද අඩු කරන්න ඕනේ.

## 2. Regression Cost Functions (ඉලක්කම් අනුමාන කරද්දී)

Regression කියන්නේ නිශ්චිත අගයක් (මිලක්, උසක්, බරක්) predict කරනවට. මෙතනදී අපි පාවිච්චි කරන ප්‍රධාන Functions කීපයක් තියෙනවා.

### 2.1 Mean Squared Error (MSE) / L2 Loss
මේක තමයි Regression වල රජා! අපි කරන්නේ ඇත්ත උත්තරෙයි predict කරපු උත්තරෙයි වෙනස අරගෙන, ඒක වර්ග කරලා (Square කරලා) ඒ ඔක්කොගෙම සාමාන්‍ය අගය ගන්න එක.

**ඇයි Square කරන්නේ?**
1. සෘණ අගයන් (negative values) ධන වෙනවා.
2. පරතරය ලොකු වෙන්න වෙන්න, Square කරද්දී දෙන දඬුවම (penalty) ගොඩක් ලොකු වෙනවා. (වෙනස 10යි නම් cost එක 100යි!). ඒ නිසා model එක ලොකු වැරදි කරන්න ගොඩක් බයයි.
3. MSE එකේ ප්‍රස්ථාරය (graph) ලස්සන U-හැඩයක් (Parabola) ගන්නවා. ඒක නිසා Gradient Descent එකට පහළට බැහැලා minimum එක හොයාගන්න ගොඩක් ලේසියි.

**අවුල:** Outliers වලට (අසාමාන්‍ය දත්ත) ගොඩක් සංවේදී නිසා එකපාරටම Cost එක ගොඩක් වැඩි වෙලා model එක වැරදි පැත්තට යන්න පුළුවන්.

### 2.2 Mean Absolute Error (MAE) / L1 Loss
මෙහිදී කරන්නේ වෙනස square කරන්නේ නැතුව ධන අගය (Absolute difference) විතරක් ගන්න එක. 
මේක outliers වලට ගොඩක් බය නැති නිසා, dataset එකේ අසාමාන්‍ය අගයන් ගොඩක් තියෙනවා නම් MAE එක හොඳයි. හැබැයි මේකේ minimum එක හොයාගන්න එක සමහර වෙලාවට අමාරුයි (හැඩය V අකුර වගේ නිසා යටම තිතේදී gradient එක හරියට ගණනය කරන්න අමාරුයි).

### 2.3 Huber Loss
මේක MAE එකෙයි MSE එකෙයි දෙමුහුමක් (Hybrid එකක්). අපිට ඕනේ MAE එකේ Outlier-resistance එකයි (outliers වලට බය නැති කමයි), MSE එකේ අර ලස්සන U හැඩයයි (gradient එක ලේසියෙන් හොයාගන්න පුළුවන් කමයි) කියන දෙකම.

**කොහොමද වැඩ කරන්නේ?**
අපි පොඩි අගයක් (delta කියලා) දෙනවා.
- වැරැද්ද (error) ඒ delta එකට වඩා අඩු නම්: මේක MSE එක වගේ වැඩ කරනවා (square කරනවා). එතකොට minimum එක හොයාගන්න ලේසියි.
- වැරැද්ද ඒ delta එකට වඩා වැඩි නම් (ඒ කියන්නේ Outlier එකක් නම්): මේක MAE එක වගේ වැඩ කරනවා (කෙලින් ඉරක් වගේ). එතකොට outlier එක නිසා loss එක එකපාරටම අහසට යන්නේ නෑ.

### 2.4 Log-Cosh Loss
මේකත් Huber Loss එක වගේමයි. පොඩි වැරදි වලදී $x^2/2$ වගේ (MSE වගේ) වැඩ කරනවා. ලොකු වැරදි වලදී $abs(x) - log(2)$ වගේ (MAE වගේ) වැඩ කරනවා. විශේෂත්වය තමයි මේක කොතනකදිවත් කඩතොළු වෙලා නෑ (twice differentiable), ඒ නිසා Huber එකටත් වඩා ගණිතමය වශයෙන් ගණනය කිරීම් කරන්න ලේසියි.

```python
# Cost Function Examples (MSE vs MAE vs Huber)
y_true = np.array([100, 150, 200, 1000]) # 1000 කියන්නේ outlier එකක්
y_pred = np.array([90, 155, 180, 260])

mae = np.mean(np.abs(y_true - y_pred))
mse = np.mean(np.square(y_true - y_pred))
huber = tf.keras.losses.Huber(delta=1.0)(y_true, y_pred).numpy()

print(f"Mean Absolute Error (MAE): {mae}")
print(f"Mean Squared Error (MSE): {mse}")
print(f"Huber Loss: {huber}")
print("දැක්කද? Outlier එකක් නිසා MSE එක MAE එකට වඩා ගොඩාක් විශාල වෙලා. හැබැයි Huber Loss එක ඒක පාලනය කරගෙන තියෙනවා.")
```

## 3. Classification Loss Functions (වර්ගීකරණය කරද්දී)

Classification වලදී අපි predict කරන්නේ ඉලක්කමක් නෙවෙයි, සම්භාවිතාවක් (Probability එකක්). එතකොට අපිට MSE වගේ දේවල් හරියන්නේ නෑ.

### 3.1 Binary Cross-Entropy (BCE)
කණ්ඩායම් (classes) දෙකක් විතරක් තියෙනකොට (බල්ලා vs පූසා) මේක පාවිච්චි කරනවා. Model එක අන්තිමට Sigmoid එකකින් 0 ත් 1 ත් අතර අගයක් දෙනවා. 
Model එක හරියටම 1.0 (ඇත්ත) කියලා predict කරොත් Loss එක 0යි. හැබැයි 0.1 කියලා predict කරලා ඇත්ත උත්තරේ 1 වුණොත්, Logarithm එක නිසා Loss එක අසීමිතව (infinitely) ලොකු වෙනවා! මේකෙන් model එකට බොරු විශ්වාසයක් තියාගන්න බැරි වෙනවා.

### 3.2 Categorical Cross-Entropy (CCE)
Classes 3ක් හෝ ඊට වඩා තියෙනවා නම් මේක පාවිච්චි කරනවා. මෙතනදී Target එක One-Hot Encoded විදිහට තියෙන්න ඕනේ (උදා: [0, 1, 0]). අවසාන layer එකට Softmax පාවිච්චි කරනවා. 
මෙතනදී Loss එක ගණනය කරන්නේ ඇත්ත class එකේ (1 තියෙන තැන) probability එක අරගෙන ඒකේ -log අගය ගැනීමෙනි. අනිත් classes වල probability එක මේ Loss එකට කෙලින්ම බලපාන්නේ නෑ (ඒවා Softmax එක ඇතුලෙදි adjust වෙනවා).

### 3.3 Sparse Categorical Cross-Entropy (SCCE)
මේකත් Categorical Cross-Entropy ම තමයි. වෙනස තියෙන්නේ අපේ Target data (y_true) වල හැඩය.
අපිට Classes 1000ක් විතර තියෙනවා නම්, `[0, 0, 0, ... 1, 0]` වගේ One-Hot array එකක් හදන එක memory එක නාස්ති කිරීමක්. 
ඒ වෙනුවට කෙලින්ම class එකේ අංකය (index එක, උදා: 2) දෙන්න පුළුවන්. ඒකට තමයි **Sparse** කියන්නේ.

## 4. Advanced / Specialized Loss Functions (තවත් වර්ග)

### 4.1 Hinge Loss
ගොඩක් වෙලාවට Support Vector Machines (SVMs) වල පාවිච්චි වෙනවා. මේකෙදී Classes තියෙන්නේ 0 සහ 1 විදිහට නෙවෙයි, -1 සහ +1 විදිහට. 
Model එකේ prediction එක හරි පැත්තේ (සෘණ හෝ ධන) ප්‍රමාණවත් තරම් ඈතින් (margin) තියෙනවා නම් Loss එක 0යි. නැත්නම් ඒ දුරට අනුව Loss එකක් එනවා. මේක Classification වලදී margins ලොකු කරන්න පාවිච්චි කරන හොඳ loss එකක්.

### 4.2 KL Divergence (Kullback-Leibler Divergence)
Probability distributions දෙකක් (සම්භාවිතා ව්‍යාප්ති දෙකක්) අතර වෙනස මනින්න පාවිච්චි කරනවා. උදාහරණයක් විදිහට VAE (Variational Autoencoders) වගේ Generative models වල ගොඩක් වැදගත් වෙනවා. ඇත්ත distribution එකෙන් අපේ model එකේ distribution එක කොච්චර ඈත් වෙලාද කියලා මනිනවා.

## 5. Which Loss Function To Use When? (මොන වෙලාවට මොකක්ද ගන්නේ?)

Loss Function එක තෝරගන්න එක හරිම ලේසියි. මේ Summary Table එක බලන්න:

| ප්‍රශ්නයේ වර්ගය (Task) | උදාහරණයක් | Output Activation | භාවිතා කළ යුතු Loss Function |
|---|---|---|---|
| **Regression (සාමාන්‍ය)** | ගෙදරක මිල කීයද? | `Linear` | `MSE` |
| **Regression (Outliers තියෙන)** | ගෙදරක මිල (එක ගෙයක් කෝටි 100යි) | `Linear` | `Huber` හෝ `MAE` |
| **Binary Classification** | මේක Spam ද නැද්ද? | `Sigmoid` | `BinaryCrossentropy` |
| **Multi-class Classification** (One-Hot target) | බල්ලෙක්ද, පූසෙක්ද, කුරුල්ලෙක්ද? | `Softmax` | `CategoricalCrossentropy` |
| **Multi-class Classification** (Integer target) | බල්ලෙක්ද(0), පූසෙක්ද(1), කුරුල්ලෙක්ද(2)? | `Softmax` | `SparseCategoricalCrossentropy` |

## 6. What is an Optimizer? (Optimizer එකක් කියන්නේ මොකක්ද?)

Cost function එකෙන් "ඔයා කරපු වැරැද්ද මෙච්චරයි" කියලා කිව්වට පස්සේ, ඒ වැරැද්ද අඩු කරගන්න Weights සහ Biases වෙනස් කරන්න ඕනේ කොයි පැත්තටද, කොච්චර ප්‍රමාණයකින්ද කියලා තීරණය කරන්නේ Optimizer එකෙන්. සරලවම කිව්වොත් මේක තමයි කන්දෙන් (Loss surface) පල්ලෙහාට බහින්න පාර පෙන්නන "මාර්ගෝපදේශකයා".

අපි මේ කොටසින් පහත සඳහන් ප්‍රධාන Optimizers ගැන ගැඹුරින් කතා කරනවා:
1. **SGD (Stochastic Gradient Descent)** - මූලිකම ක්‍රමය
2. **Mini-Batch SGD** - ප්‍රායෝගිකව ගොඩක් පාවිච්චි වෙන ක්‍රමය
3. **SGD with Momentum** - බෝලයක් කන්දෙන් පෙරලනවා වගේ වේගවත් ක්‍රමය
4. **Adagrad** - හැම weight එකකටම වෙනම learning rate එකක් දෙන ක්‍රමය
5. **RMSprop** - Adagrad එකේ අඩුපාඩු හදපු ක්‍රමය
6. **Adam** - අද කාලේ හැමෝම පාවිච්චි කරන ජනප්‍රියම ක්‍රමය (Momentum + RMSprop)

## 7. SGD (Stochastic Gradient Descent)


**Batch Gradient Descent:** 
මේකෙදී කරන්නේ මුළු Dataset එකම එකපාර Network එක ඇතුලට දාලා, සම්පූර්ණ Cost එක හොයලා, එක පාරක් Weights update කරන එක. Dataset එක ලොකු නම් (උදා: මිලියන 10ක් images), එක step එකක් යන්න ගොඩක් වෙලා යනවා. Memory එකත් මදි වෙනවා.

**SGD (Stochastic Gradient Descent):**
Stochastic කියන්නේ "අහඹු" (Random) කියන එකයි. මෙතනදී කරන්නේ මුළු dataset එකම ගන්නේ නැතුව, **එක පාරකට එක training example එකක්** විතරක් අරගෙන ඒකෙන් Loss එක හොයලා Weights update කරන එක.

**වාසි:**
- ගොඩාක් වේගවත්.
- Memory එක යන්නේ නෑ.

**අවාසි:**
- එක දත්තයක් විතරක් අරන් තීරණ ගන්න නිසා, කන්දෙන් පහළට බහිද්දී කෙලින්ම යන්නේ නැතුව වටේ ගිහින් (Noisy/zigzag path) තමයි පල්ලෙහාට යන්නේ. මේක නිසා හරියටම minimum එකේ නවතින්නේ නැතුව එතන වටේ කැරකි කැරකි ඉන්න පුළුවන්.

## 8. Mini Batch With SGD

Batch Gradient Descent (ඔක්කොම එකපාර) ගොඩක් හෙමින්. SGD (එකින් එක) ගොඩක් Noisy. 
මේ දෙකේම මැද මාවත තමයි **Mini-Batch SGD**.

අපි කරන්නේ මුළු dataset එකම පොඩි පොඩි කෑලි වලට (Batches) කඩන එක. උදාහරණයක් විදිහට Batch Size එක 32ක් විදිහට ගන්නවා. 
එතකොට පින්තූර 32ක් එකපාර Network එකට දාලා, ඒ 32න් average Cost එක අරගෙන Weights update කරනවා. ඊළඟට ඊළඟ 32 ගන්නවා.

**වාසි:**
1. Vectorization (Matrix multiplication) පාවිච්චි කරන්න පුළුවන් නිසා GPU එකෙන් උපරිම වැඩ ගන්න පුළුවන්.
2. තනි දත්තයකට වඩා දත්ත 32ක සාමාන්‍යය ගන්න නිසා SGD එකට වඩා ගමන් මග (path) පැහැදිලියි (Less noisy).
3. Deep Learning වල standard ක්‍රමය මේක තමයි! (සාමාන්‍යයෙන් batch sizes: 32, 64, 128 පාවිච්චි කරනවා).

## 9. SGD With Momentum

SGD හෝ Mini-Batch SGD වල තියෙන ලොකුම ප්‍රශ්නයක් තමයි කන්දෙන් (Loss surface) පල්ලෙහාට බහිද්දී සමහර තැන් වලදී වමටයි දකුණටයි පැන පැන (oscillations) හෙමින් පහළට යන එක.

**Momentum කියන්නේ මොකක්ද?**
හිතන්න ඔයා කන්දක් උඩ ඉඳන් බෝලයක් අතහරිනවා කියලා. බෝලය පල්ලෙහාට යන්න යන්න ඒකේ වේගය (momentum) වැඩි වෙනවා නේද? පොඩි පොඩි වලවල් තිබ්බත් ඒ වේගය නිසා බෝලය ඒවගෙන් පැනලා යනවා. SGD with Momentum කියන්නෙත් අන්න ඒ සංකල්පයම තමයි.

**ඒක වැඩ කරන්නේ කොහොමද?**
සාමාන්‍ය SGD වලදී අපි අලුත් gradient එක දිහා විතරයි බලන්නේ. හැබැයි Momentum වලදී අපි **කලින් ගිය steps වල gradients වල සාමාන්‍යයක් (Moving Average)** මතක තියාගන්නවා. 

සමීකරණය:
`Velocity = (beta * old_Velocity) + ((1 - beta) * current_gradient)`
`New_Weight = Old_Weight - (Learning_Rate * Velocity)`

මෙහි `beta` (සාමාන්‍යයෙන් 0.9) කියන්නේ කලින් ගිය වේගය කොච්චර මතක තියාගන්නවද කියන එක. 
මේක නිසා වමටයි දකුණටයි යන අනවශ්‍ය චලනයන් කැපිලා ගිහින්, කෙලින්ම පහළට යන වේගය වැඩි වෙනවා. මේක සාමාන්‍ය SGD එකට වඩා ගොඩක් වේගවත් වගේම Local Minima වල හිරවෙන එකත් අඩු කරනවා.

## 10. Advanced Optimizers (SGD වලින් එහාට)

අපි කලින් SGD සහ Momentum ගැන කතා කළා. SGD වල තියෙන එක ලොකු අවුලක් තමයි, **හැම Weight එකකටම තියෙන්නේ එකම Learning Rate එකක් (අඩි තියන ප්‍රමාණය) වීම**. 
හිතන්න ඔයාට weights 100ක් තියෙනවා කියලා. සමහර weights වලට ගොඩක් ලොකු වෙනස්කම් ඕනේ. සමහර weights වලට ඕනේ ගොඩක් පොඩි වෙනස්කම් (fine-tuning). මේකට විසඳුමක් විදිහට තමයි "Adaptive" Learning Rate තියෙන Optimizers ආවේ.
### 9.1 Adagrad (Adaptive Gradient Algorithm)
Adagrad කරන්නේ හැම weight එකකටම වෙන වෙනම learning rate එකක් දෙන එකයි.

**කොහොමද කරන්නේ?**
මේකෙදී අපි කලින් ගියපු හැම පියවරකදීම ආපු Gradients වල වර්ග (Squares) එකතු කරගෙන යනවා. 
යම්කිසි weight එකක් ගොඩක් වෙනස් වෙලා (update වෙලා) තියෙනවා නම්, Adagrad එක ඒ weight එකේ Learning Rate එක අඩු කරනවා (හේතුව: එයා දැනටමත් ගොඩක් දුර ගිහින් ඉන්නේ, දැන් හිමින් යන්න ඕනේ).
යම්කිසි weight එකක් වැඩිය update වෙලා නැත්නම්, ඒකේ Learning Rate එක වැඩි කරනවා.

**අවුල:**
කාලය යන්න යන්න Gradients වල වර්ග එකතු වෙන නිසා, යටින් තියෙන අගය (denominator) ගොඩක් ලොකු වෙනවා. ඒ නිසා Learning Rate එක එන්න එන්නම පොඩි වෙලා අන්තිමට 0ටම කිට්ටු වෙනවා. එතකොට model එක තවත් ඉගෙනගන්න එක (learning) නවතිනවා.

### 9.2 RMSprop (Root Mean Square Propagation)
RMSprop කියන්නේ Adagrad එකේ අර තිබ්බ ලොකුම අවුල (Learning Rate එක 0 වෙන එක) විසඳන්න ආපු ක්‍රමයක්.

**වෙනස මොකක්ද?**
Adagrad වල වගේ මුල ඉඳන්ම ඔක්කොම Gradients වල වර්ග එකතු කරන්නේ නෑ. ඒ වෙනුවට මේකෙදීත් Momentum වගේම **Moving Average** එකක් පාවිච්චි කරනවා. ඒ කියන්නේ ළඟදී ආපු Gradients වලට විතරක් ලොකු බරක් (weight) දෙනවා, පරණ ඒවා අමතක කරලා දානවා.

මේක නිසා Learning Rate එක එකපාරටම 0 වෙලා learning නවතින්නේ නෑ. කාලයක් යනකන් හොඳට train වෙනවා.

### 9.3 Adam (Adaptive Moment Estimation)
මේක තමයි **Deep Learning වල රජා**! 90% ක්ම වෙලාවට හැමෝම default විදිහට පාවිච්චි කරන්නේ Adam Optimizer එක.

Adam කියන්නේ සරලවම **Momentum + RMSprop** වල එකතුවක්. මේකෙන් අර ක්‍රම දෙකේම හොඳ ලක්ෂණ එකට එකතු කරලා තියෙනවා.

1. **Momentum එකෙන් (First Moment):** කන්දෙන් පහළට යන වේගය (velocity) මතක තියාගෙන, වමටයි දකුණටයි පනින එක (oscillations) අඩු කරලා කෙලින්ම පහළට යන්න උදව් කරනවා.
2. **RMSprop එකෙන් (Second Moment):** හැම weight එකකටම වෙන වෙනම, ළඟදී ආපු gradients අනුව (moving average of squared gradients) Learning Rate එක adjust කරනවා.

**වාසි:**
- ගොඩක් වේගවත්.
- Default learning rate (0.001) එක ගොඩක් වෙලාවට කිසිම වෙනසක් කරන්නේ නැතුවම හොඳට වැඩ කරනවා.
- Memory එක යන්නේ අඩුවෙන්.
